```python
import pandas as pd
import subprocess
import re
```

```python
# run a command in shell and get the output
def shell(command):
	return subprocess.getoutput(command)

# binary_file = "/home/prapti/RA_Kate/dis-viz/analysis/correspondence_test/build/raja-perf-g-O3.exe"
binary_file = "/home/prapti/RA_Kate/dis-viz/analysis/correspondence_test/build/test"
command = f"dwarfdump -l {binary_file}"
output = shell(command).split("\n")

# remove all lines that does not start with an address
output = [line for line in output if line.startswith("0x")]
output
```

```python
# number of duplicate lines
duplicates = len(output) - len(set(output))
print(f"Number of duplicate lines: {duplicates}")
```

```python
prev_source_file = ""
data = []
line_regex = re.compile(
    r'(?P<address>0x[0-9a-f]+)\s*\[\s*(?P<line>\d+),\s*(?P<column>\d+)\]\s*NS(?:\s+uri:\s*"(?P<file>[^"]+)")?'
)
for line in output:
	match = line_regex.match(line)
	if match:
		pc, lno, col, source_file = match.groups()
		data.append({
			"pc": int(pc, 16),
			"lno": int(lno),
			"col": int(col),
			"source_file": source_file.strip("\"") if source_file else prev_source_file,
		})
		if source_file:
			prev_source_file = source_file.strip("\"")
	else:
		print(f"Failed to parse: {line}")
		
df = pd.DataFrame(data)
# remove duplicate rows
df
```

```python
#check if df has duplicates in pc

duplicates = df[df.duplicated(subset=["pc"], keep=False)]
duplicates
```

```python
# print in this format: 0x11ed: mov %rdx,%rsi   (bubble_sort.cpp:10)
for _, row in df.iterrows():
	print(f"{hex(row['pc'])}:\t({row['source_file'].split('/')[-1]}:{row['lno']})")
```

```python
# parse the output
# <pc>        [lno,col] NS BB ET PE EB IS= DI= uri: "filepath"
# '0x00001169  [   8, 1] NS uri: "/home/prapti/RA_Kate/dis-viz/sample_inputs/bubble_sort.cpp"'
# '0x000011f8  [  14, 9] NS DI=0x2'
# '0x000013a3  [ 200, 5] NS ET'
data = []
line_regex = re.compile(r"0x([0-9a-f]+)\s+\[\s*(\d+),\s*(\d+)\].*")
for line in output:
	match = line_regex.match(line)
	if match:
		pc, lno, col = match.groups()
		data.append({
			"pc": int(pc, 16),
			"lno": int(lno),
			"col": int(col),
		})
	else:
		print(f"Failed to parse: {line}")

df = pd.DataFrame(data)
df.head()
```
